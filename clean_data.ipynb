## Yield Data

```python
poly = gpd.read_file('D:/Work/ADB/SAR/gadm41_VNM_1.json')
poly['VARNAME_1'] = poly['VARNAME_1'].apply(lambda x: re.sub(r"(\w)([A-Z])", r"\1 \2",x ))
poly['VARNAME_1'] = poly['VARNAME_1'].str.replace('-', ' - ')
poly = poly[['VARNAME_1']]
poly = poly.rename({'VARNAME_1': 'Province'}, axis=1)
poly
```

```python
provinces = poly['Province'].to_list()
```

```python
df_total = pd.read_excel(os.path.join(local, 'Total paddy yield.xlsm'))
# df_total = df_total.replace(0, np.nan)
df_total
```

```python
paths = ['Total paddy yield.xlsm', 'Spring paddy yield.xlsm',
         'Autumn paddy yield.xlsm', 'Winter paddy yield.xlsm']

for p in paths:
    df = pd.read_excel(os.path.join(local, p))
    df = df.replace(0, np.nan)
    df['Province'] = df['Province'].str.replace('  ', ' ')
    df.to_csv(os.path.join(local, p).replace('xlsm', 'csv'), index=False)
    
```

```python
df_a = pd.read_csv(os.path.join(local, 'Autumn paddy yield.csv'))
df_a = pd.merge(poly, df_a, on=['Province'], how='left')
df_a
```

```python
df_a.to_csv(os.path.join(local, 'Autumn paddy yield.csv'), index=False)
```

## Leaf based

```python
df = pd.read_csv(os.path.join(local, 'LEAF_NORTH.csv'), dtype={'lat': 'str',
                                                                'lon': 'str'})
coords = pd.read_csv('datasets/province_by_coords.csv', dtype={'lat': 'str',
                                                                'lon': 'str'})
```

```python
df = pd.merge(df, coords, on=['lon', 'lat'], how='left')
df['year'] = df['Timestamp'].str[:4]
df = df.drop(['lat', 'lon', 'Timestamp'], axis=1)
df
```

```python
gb = df.groupby(['Province', 'year']).agg({'LAI_mean': 'mean',
                                           'NDVI_mean': 'mean'}).reset_index()

# gb = df.groupby(['Province', 'year']).agg({'VV': 'mean',
#                                            'VH': 'mean',
#                                            'VV/VH': 'mean'}).reset_index()

gb
                                            
```

```python
gb.to_csv('datasets/aggregated/LEAF_GB_NORTH.csv', index=False)
```
