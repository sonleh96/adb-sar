```python
def assign_municipalities(df, poly, col_name):
    localdf = df[['lon', 'lat']].copy(deep=True)
    try:
        local_gdf = gpd.GeoDataFrame(
            localdf, crs='epsg:4326',
            geometry=[Point(xy) for xy in zip(localdf['lon'], localdf['lat'])]
        )
        local_gdf = gpd.sjoin(
            local_gdf, poly, how='left', op='within')
        
        if col_name in local_gdf.columns:
            return local_gdf[col_name].reset_index(drop=True)
        else:
            raise ValueError(f"Column '{col_name}' not found in the joined GeoDataFrame.")
        
    except ValueError as ve:
        print(ve)
        series = pd.Series([np.nan] * len(localdf))
        return series

def create_bounding_box(df, center_lon_col, center_lat_col, side_length_m):
    
    side_length_deg = side_length_m /111000

    half_side = side_length_deg / 2

    # Calculate bounding box corners
    df['min_lon'] = df[center_lon_col] - half_side
    df['max_lon'] = df[center_lon_col] + half_side
    df['min_lat'] = df[center_lat_col] - half_side
    df['max_lat'] = df[center_lat_col] + half_side
    
    return df[['lon', 'lat', 'min_lon', 'min_lat', 'max_lon', 'max_lat']]
```

```python
from shapely.geometry import Polygon

df_bbox = dd.read_csv(r"D:\Work\ADB\SAR\datasets\VN_SOUTH_VER2_bbox.csv")
def create_polygon_from_bbox(row):
    # Extract bounding box coordinates
    # Define the four corners of the bounding box in the order needed for a Polygon
    return Polygon([(row['min_lon'], row['min_lat']), (row['min_lon'], row['max_lat']), (row['max_lon'], row['max_lat']), (row['max_lon'], row['min_lat']), (row['min_lon'], row['min_lat'])])

df_bbox['geometry'] = df_bbox.map_partitions(lambda df: df.apply(create_polygon_from_bbox, axis=1), meta=('polygon', 'object'))
df_bbox[['lon', 'lat', 'geometry']].to_csv(r"D:\Work\ADB\SAR\datasets\VN_SOUTH_VER2_bbox.csv", index=False, single_file = True)

# gdf = gpd.GeoDataFrame(df_bbox[['lon', 'lat']],
#                        crs='epsg:4326',
#                        geometry=df_bbox['polygon'])
```

```python
import pandas as pd
from shapely.geometry import Polygon
df = pd.read_csv(r"D:\Work\ADB\SAR\datasets\VN_SOUTH_VER2_bbox.csv")
def bbox_to_polygon(row):
    return Polygon([
        (row['min_lon'], row['min_lat']),
        (row['min_lon'], row['max_lat']),
        (row['max_lon'], row['max_lat']),
        (row['max_lon'], row['min_lat']),
        (row['min_lon'], row['min_lat'])  # Close the polygon
    ])

# Apply the function to create a 'geometry' column with polygons
df['geometry'] = df.apply(bbox_to_polygon, axis=1)

# Convert DataFrame to GeoDataFrame with CRS (Coordinate Reference System)
gdf = gpd.GeoDataFrame(df, geometry='geometry', crs="EPSG:4326")

# Drop the original bounding box columns if needed
gdf = gdf.drop(columns=['min_lon', 'min_lat', 'max_lon', 'max_lat'])
```
